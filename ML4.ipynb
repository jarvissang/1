```python
# 1. Define the function and its derivative
def f(x):
    return (x + 3)**2

def df(x):
    # The derivative of (x+3)^2 is 2(x+3)
    return 2 * (x + 3)

# 2. Initialize parameters
current_x = 2.0          # Starting point as given in the problem
learning_rate = 0.1      # Step size (how big of a jump we take)
iterations = 35          # Number of steps to take

# 3. Implement Gradient Descent
print(f"Starting at x = {current_x}, y = {f(current_x)}")
print("-" * 30)

for i in range(iterations):
    # Calculate the gradient (slope) at the current point
    gradient = df(current_x)

    # Update x by moving in the opposite direction of the gradient
    current_x = current_x - (learning_rate * gradient)

    # Print progress
    print(f"Iteration {i+1:2}: x = {current_x:.4f}, y = {f(current_x):.4f}")

print("-" * 30)
print(f"Local minimum found at x ≈ {current_x:.4f}")
```
